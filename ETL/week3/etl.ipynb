```python
print("=== DATA CONSISTENCY CHECKS ===")

# Check for data inconsistencies and clean them
consistency_issues = []

# Clean numerical columns that might have string formatting (commas, etc.)
numerical_string_cols = []
for col in df_transformed.columns:
    if df_transformed[col].dtype == 'object':
        # Check if column contains numerical data stored as strings
        sample_values = df_transformed[col].dropna().head(10)
        if any(str(val).replace(',', '').replace('.', '').isdigit() for val in sample_values):
            numerical_string_cols.append(col)

print(f"Columns with numerical data stored as strings: {numerical_string_cols}")

# Convert numerical string columns to proper numerical format
for col in numerical_string_cols:
    try:
        # Remove commas and convert to numeric
        df_transformed[col] = pd.to_numeric(df_transformed[col].astype(str).str.replace(',', ''), errors='coerce')
        print(f"✓ Converted '{col}' from string to numerical format")
    except Exception as e:
        print(f"⚠ Could not convert '{col}': {e}")

# Check for negative values in columns that shouldn't have them (like streams, counts)
stream_columns = [col for col in df_transformed.columns if 'stream' in col.lower() or 'count' in col.lower() or 'spin' in col.lower()]
for col in stream_columns:
    if df_transformed[col].dtype in ['int64', 'float64']:
        negative_count = (df_transformed[col] < 0).sum()
        if negative_count > 0:
            print(f"⚠ Found {negative_count} negative values in '{col}' - these will be set to 0")
            df_transformed[col] = df_transformed[col].clip(lower=0)

print("\n✓ Data consistency checks completed!")
```

## 6. Normalization and Standardization


```python
# 6.1 Apply normalization and standardization
print("=== NORMALIZATION AND STANDARDIZATION ===")

# Create final dataset for modeling
df_final = df_typed.copy()

# 6.1.1 Identify features that need scaling
print("\n1. Analyzing feature scales...")

scaling_candidates = []
for feature in available_predictors:
    if df_final[feature].dtype in ['int64', 'float64']:
        feature_std = df_final[feature].std()
        feature_mean = df_final[feature].mean()
        feature_range = df_final[feature].max() - df_final[feature].min()
        
        print(f"{feature}:")
        print(f"  Range: {df_final[feature].min():,.0f} to {df_final[feature].max():,.0f}")
        print(f"  Mean: {feature_mean:,.1f}, Std: {feature_std:,.1f}")
        
        # Add to scaling candidates if range is large or if it's a count/stream variable
        if feature_range > 1000 or 'stream' in feature.lower() or 'count' in feature.lower():
            scaling_candidates.append(feature)
            print(f"  → Will be scaled (large range or count variable)")
        else:
            print(f"  → No scaling needed")
        print()

print(f"Features selected for scaling: {len(scaling_candidates)}")
for feature in scaling_candidates:
    print(f"  • {feature}")

# 6.1.2 Apply StandardScaler
print(f"\n2. Applying StandardScaler...")
print("Justification: StandardScaler is chosen because:")
print("• Preserves the shape of the original distribution")
print("• Handles outliers better than Min-Max scaling")
print("• Suitable for features with different units (streams, counts, rankings)")
print("• Required for many ML algorithms (SVM, Neural Networks, etc.)")

if scaling_candidates:
    scaler = StandardScaler()
    
    # Create scaled versions of the features
    for feature in scaling_candidates:
        original_col = feature
        scaled_col = f"{feature}_scaled"
        
        # Fit and transform the feature
        df_final[scaled_col] = scaler.fit_transform(df_final[[original_col]])
        
        print(f"✓ Scaled '{original_col}' → '{scaled_col}'")
        print(f"   Original range: {df_final[original_col].min():,.0f} to {df_final[original_col].max():,.0f}")
        print(f"   Scaled range: {df_final[scaled_col].min():.3f} to {df_final[scaled_col].max():.3f}")
        print(f"   Scaled mean: {df_final[scaled_col].mean():.3f}, Scaled std: {df_final[scaled_col].std():.3f}")
        print()
    
    # Update predictor list to use scaled versions
    scaled_predictors = []
    for predictor in available_predictors:
        if predictor in scaling_candidates:
            scaled_predictors.append(f"{predictor}_scaled")
        else:
            scaled_predictors.append(predictor)
    
    print(f"✓ Standardization completed!")
    print(f"Updated predictor list with scaled features: {len(scaled_predictors)} features")
    
else:
    scaled_predictors = available_predictors
    print("No features required scaling.")

# 6.1.3 Final dataset summary
print(f"\n=== FINAL TRANSFORMATION SUMMARY ===")
print(f"Final dataset shape: {df_final.shape}")
print(f"Target variable: {target_variable}")
print(f"Number of predictors: {len(scaled_predictors)}")
print(f"Features ready for modeling: {len(scaled_predictors) + 1}")  # +1 for target
```

## 7. Additional Transformations


```python
# 7.1 Additional domain-specific transformations
print("=== ADDITIONAL TRANSFORMATIONS ===")

# 7.1.1 Log Transformations for Highly Skewed Data
print("\n1. Log Transformations for Skewed Streaming Data")
print("Justification: Streaming data often follows power-law distributions with extreme outliers.")
print("Log transformation helps normalize these distributions for better model performance.")

# Identify highly skewed streaming features
skewed_features = []
log_features = []

for feature in ['Spotify Streams', 'YouTube Views', 'TikTok Views', 'Pandora Streams', 'Soundcloud Streams']:
    if feature in df_final.columns:
        # Calculate skewness
        from scipy import stats
        skewness = stats.skew(df_final[feature].dropna())
        
        print(f"\n{feature}:")
        print(f"  Skewness: {skewness:.2f}")
        
        if abs(skewness) > 2:  # Highly skewed
            print(f"  → Highly skewed, applying log transformation")
            # Add 1 to handle zero values
            df_final[f'{feature}_log'] = np.log1p(df_final[feature])
            log_features.append(f'{feature}_log')
            skewed_features.append(feature)
        else:
            print(f"  → Not highly skewed, no transformation needed")

print(f"\nCreated log-transformed features: {log_features}")

# 7.1.2 Ratio Features for Cross-Platform Analysis
print(f"\n2. Cross-Platform Ratio Features")
print("Justification: Ratios help identify platform preferences and engagement patterns.")

ratio_features = []

# Engagement ratios
if 'YouTube Views' in df_final.columns and 'YouTube Likes' in df_final.columns:
    df_final['youtube_engagement_ratio'] = (
        df_final['YouTube Likes'] / (df_final['YouTube Views'] + 1)
    ) * 100  # Percentage
    ratio_features.append('youtube_engagement_ratio')
    print("✓ Created YouTube engagement ratio (likes/views)")

if 'TikTok Views' in df_final.columns and 'TikTok Likes' in df_final.columns:
    df_final['tiktok_engagement_ratio'] = (
        df_final['TikTok Likes'] / (df_final['TikTok Views'] + 1)
    ) * 100
    ratio_features.append('tiktok_engagement_ratio')
    print("✓ Created TikTok engagement ratio (likes/views)")

# Platform dominance ratios
if 'Spotify Streams' in df_final.columns and 'YouTube Views' in df_final.columns:
    total_main_streams = df_final['Spotify Streams'] + df_final['YouTube Views']
    df_final['spotify_dominance'] = df_final['Spotify Streams'] / (total_main_streams + 1)
    ratio_features.append('spotify_dominance')
    print("✓ Created Spotify dominance ratio")

print(f"Created ratio features: {ratio_features}")

# 7.1.3 Binning for Categorical Insights
print(f"\n3. Binning Continuous Variables")
print("Justification: Binning helps capture non-linear relationships and creates interpretable categories.")

binned_features = []

# Bin release year into decades/eras
if 'release_year' in df_final.columns:
    df_final['music_era'] = pd.cut(
        df_final['release_year'], 
        bins=[2019, 2021, 2022, 2023, 2025], 
        labels=['Pre-2021', '2021', '2022', '2023+'],
        include_lowest=True
    )
    binned_features.append('music_era')
    print("✓ Created music era categories from release year")

# Bin track score into performance tiers
if 'Track Score' in df_final.columns:
    score_bins = [0, 200, 400, 600, float('inf')]
    score_labels = ['Low', 'Medium', 'High', 'Viral']
    df_final['performance_tier'] = pd.cut(
        df_final['Track Score'], 
        bins=score_bins, 
        labels=score_labels,
        include_lowest=True
    )
    binned_features.append('performance_tier')
    print("✓ Created performance tier categories from Track Score")

print(f"Created binned features: {binned_features}")

# 7.1.4 Interaction Features
print(f"\n4. Interaction Features")
print("Justification: Interaction features capture synergistic effects between platforms.")

interaction_features = []

# Social media synergy
if 'TikTok Posts' in df_final.columns and 'YouTube Views' in df_final.columns:
    df_final['social_media_synergy'] = df_final['TikTok Posts'] * np.log1p(df_final['YouTube Views'])
    interaction_features.append('social_media_synergy')
    print("✓ Created social media synergy (TikTok × YouTube)")

print(f"Created interaction features: {interaction_features}")

# 7.1.5 Final feature summary
all_new_features = log_features + ratio_features + binned_features + interaction_features
print(f"\n=== ADDITIONAL TRANSFORMATIONS SUMMARY ===")
print(f"Total new features created: {len(all_new_features)}")
print(f"Log-transformed features: {len(log_features)}")
print(f"Ratio features: {len(ratio_features)}")
print(f"Binned features: {len(binned_features)}")
print(f"Interaction features: {len(interaction_features)}")
print(f"\nFinal dataset shape: {df_final.shape}")

# Export transformation summary
transformations_applied = {
    'data_cleaning': 'Missing values imputed, duplicates removed, consistency checks',
    'encoding': f'Categorical variables encoded using Label/One-Hot encoding',
    'feature_engineering': f'Created {len(all_new_features)} new features',
    'data_type_conversion': 'Release Date → datetime, streaming columns → int',
    'scaling': f'StandardScaler applied to {len(scaling_candidates)} features',
    'log_transformation': f'Applied to {len(skewed_features)} highly skewed features',
    'additional_features': f'{len(ratio_features)} ratios, {len(interaction_features)} interactions'
}

print(f"\n=== COMPLETE TRANSFORMATION PIPELINE SUMMARY ===")
for step, description in transformations_applied.items():
    print(f"✓ {step.replace('_', ' ').title()}: {description}")
```

## 2. Encoding Categorical Variables


```python
# 2.1 Identify categorical variables
print("=== CATEGORICAL VARIABLE ENCODING ===")

# Identify categorical columns
categorical_cols = df_transformed.select_dtypes(include=['object']).columns.tolist()
print(f"Categorical columns found: {categorical_cols}")

# Analyze cardinality of categorical variables to choose appropriate encoding
print("\nCardinality analysis for encoding strategy:")
for col in categorical_cols:
    unique_count = df_transformed[col].nunique()
    total_rows = len(df_transformed)
    cardinality_ratio = unique_count / total_rows
    print(f"  {col}: {unique_count} unique values ({cardinality_ratio:.2%} of total)")
    
    # Show some sample values
    sample_values = df_transformed[col].value_counts().head(5)
    print(f"    Top values: {list(sample_values.index)}")
    print()

# Strategy: 
# - High cardinality (>50% unique): Consider Label Encoding or dropping if not useful
# - Low cardinality (<10 unique): One-Hot Encoding
# - Medium cardinality (10-50 unique): Label Encoding
```

```python
# 2.2 Apply appropriate encoding strategies
print("=== APPLYING ENCODING STRATEGIES ===")

# Create a copy to preserve original categorical data
df_encoded = df_transformed.copy()

# Track encoding information
encoding_info = {}

for col in categorical_cols:
    unique_count = df_encoded[col].nunique()
    cardinality_ratio = unique_count / len(df_encoded)
    
    if cardinality_ratio > 0.5:  # High cardinality (>50% unique values)
        print(f"\n'{col}': High cardinality ({unique_count} unique) - Using Label Encoding")
        le = LabelEncoder()
        df_encoded[f'{col}_encoded'] = le.fit_transform(df_encoded[col].astype(str))
        encoding_info[col] = {'method': 'Label Encoding', 'encoder': le, 'new_column': f'{col}_encoded'}
        
    elif unique_count <= 10:  # Low cardinality - One-Hot Encoding
        print(f"\n'{col}': Low cardinality ({unique_count} unique) - Using One-Hot Encoding")
        # Create dummy variables
        dummies = pd.get_dummies(df_encoded[col], prefix=col, dummy_na=False)
        df_encoded = pd.concat([df_encoded, dummies], axis=1)
        encoding_info[col] = {'method': 'One-Hot Encoding', 'new_columns': dummies.columns.tolist()}
        
    else:  # Medium cardinality - Label Encoding
        print(f"\n'{col}': Medium cardinality ({unique_count} unique) - Using Label Encoding")
        le = LabelEncoder()
        df_encoded[f'{col}_encoded'] = le.fit_transform(df_encoded[col].astype(str))
        encoding_info[col] = {'method': 'Label Encoding', 'encoder': le, 'new_column': f'{col}_encoded'}

print(f"\n✓ Encoding completed! Dataset shape: {df_encoded.shape}")
print(f"Original shape: {df_transformed.shape}")
print(f"New columns added: {df_encoded.shape[1] - df_transformed.shape[1]}")
```

## 3. Feature Engineering


```python
# 3.1 Create meaningful features for music streaming analysis
print("=== FEATURE ENGINEERING ===")

df_featured = df_encoded.copy()

# 3.1.1 Streaming Performance Features
print("\n1. Creating streaming performance features...")

# Total Streams Across Platforms (if multiple streaming columns exist)
streaming_cols = [col for col in df_featured.columns if 'stream' in col.lower() and df_featured[col].dtype in ['int64', 'float64']]
if len(streaming_cols) > 1:
    df_featured['total_streams'] = df_featured[streaming_cols].sum(axis=1)
    print(f"✓ Created 'total_streams' from: {streaming_cols}")

# 3.1.2 Playlist Engagement Features
print("\n2. Creating playlist engagement features...")

playlist_count_cols = [col for col in df_featured.columns if 'playlist count' in col.lower()]
playlist_reach_cols = [col for col in df_featured.columns if 'playlist reach' in col.lower()]

if playlist_count_cols and playlist_reach_cols:
    # Average reach per playlist
    for count_col, reach_col in zip(playlist_count_cols, playlist_reach_cols):
        platform = count_col.split()[0].lower()
        df_featured[f'{platform}_avg_reach_per_playlist'] = (
            df_featured[reach_col] / (df_featured[count_col] + 1)  # +1 to avoid division by zero
        )
        print(f"✓ Created '{platform}_avg_reach_per_playlist'")

# 3.1.3 Track Popularity Score (if Track Score exists)
if 'Track Score' in df_featured.columns:
    print("\n3. Creating popularity categories...")
    # Create popularity tiers based on Track Score
    df_featured['popularity_tier'] = pd.cut(
        df_featured['Track Score'], 
        bins=[0, 100, 300, 500, float('inf')], 
        labels=['Low', 'Medium', 'High', 'Viral'],
        include_lowest=True
    )
    print("✓ Created 'popularity_tier' based on Track Score")

# 3.1.4 Release Timing Features (if Release Date exists)
if 'Release Date' in df_featured.columns:
    print("\n4. Creating release timing features...")
    try:
        # Convert Release Date to datetime
        df_featured['release_date_parsed'] = pd.to_datetime(df_featured['Release Date'], errors='coerce')
        
        # Extract temporal features
        df_featured['release_year'] = df_featured['release_date_parsed'].dt.year
        df_featured['release_month'] = df_featured['release_date_parsed'].dt.month
        df_featured['release_quarter'] = df_featured['release_date_parsed'].dt.quarter
        df_featured['release_day_of_week'] = df_featured['release_date_parsed'].dt.dayofweek
        
        # Days since release (from latest date in dataset)
        max_date = df_featured['release_date_parsed'].max()
        df_featured['days_since_release'] = (max_date - df_featured['release_date_parsed']).dt.days
        
        print("✓ Created temporal features: year, month, quarter, day_of_week, days_since_release")
    except Exception as e:
        print(f"⚠ Could not parse release dates: {e}")

print(f"\n✓ Feature engineering completed! New shape: {df_featured.shape}")
print(f"Features added: {df_featured.shape[1] - df_encoded.shape[1]}")
```
